# Statistical Analysis for Business Decisions

```python
# Conduct hypothesis testing (t-test, chi-square)

import pandas as pd
from scipy import stats

# 1. Loading dataset 
df = pd.read_csv(r"C:\Users\Admin\Downloads\churn-bigml-80.csv")

# 2. Ensure the binary target column exists
if 'Churn_binary' not in df.columns:
    if df['Churn'].dtype == 'object':
        df['Churn_binary'] = df['Churn'].map({'True': 1, 'False': 0, 'Yes': 1, 'No': 0, True: 1, False: 0})
    else:
        df['Churn_binary'] = df['Churn'].astype(int)

# 3. Independent T-Test: Total Day Minutes vs Churn
churned_mins = df[df['Churn_binary'] == 1]['Total day minutes']
retained_mins = df[df['Churn_binary'] == 0]['Total day minutes']

t_stat, p_value_t = stats.ttest_ind(churned_mins, retained_mins, equal_var=False)
print(f"T-Test -> T-statistic: {t_stat:.4f}, p-value: {p_value_t:.4e}")
```

```python
# Using probability distributions for risk analysis

from scipy.stats import binom

# Given the overall historical churn rate (e.g., ~14.5% or 0.145)
p_churn = df['Churn_binary'].mean()
n_customers = 100  # Size of a sample customer cohort

# What is the probability of losing exactly 20 or more customers out of 100?
prob_20_or_more = 1 - binom.cdf(19, n_customers, p_churn)
print(f"Probability of 20+ churns in a cohort of 100: {prob_20_or_more:.2%}")
```

```python
# Performing A/B testing for marketing campaigns

from statsmodels.stats.proportion import proportions_ztest

# Simulated A/B test results
successes = [45, 70]   # Retained customers in Control vs Treatment
nobs = [500, 500]      # Total customers in each group

z_stat, p_val_ab = proportions_ztest(successes, nobs)
print(f"A/B Test Z-statistic: {z_stat:.4f}, p-value: {p_val_ab:.4f}")
```

```python
# Calculating confidence intervals and margin of error

import statsmodels.stats.proportion as smp

n_churn = df['Churn_binary'].sum()
n_total = len(df)

# Calculate 95% confidence interval for the true churn rate
ci_low, ci_upp = smp.proportion_confint(n_churn, n_total, alpha=0.05, method='normal')
margin_error = (ci_upp - ci_low) / 2

print(f"Estimated Churn Rate: {n_churn/n_total:.2%}")
print(f"95% Confidence Interval: [{ci_low:.2%}, {ci_upp:.2%}]")
print(f"Margin of Error: ±{margin_error:.2%}")
```
